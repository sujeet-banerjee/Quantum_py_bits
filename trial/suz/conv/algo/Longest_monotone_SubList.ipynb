```python
def longestMonotoneSublist(nums: [int]) -> int:
    '''
    It has three boundary issues!
    :param nums:
    :return:
    '''
    max_window_size = 0
    left = 0
    for right, num in enumerate(nums):
        # If num > nums[right-1] continue!
        # Else,
        ###! Incorrect, at start righ=0,
        #### thus you are comparing with the last number!
        if num < nums[right-1]:
            window_size = right - left
            if window_size > max_window_size:
                max_window_size = window_size
                ###! This must execute even if there was a
                #### smaller window (outside this if-block)
                left = right

    ###! max_window_size could be 0, if the entire list is
    #### is a monotonic sequence!
    return max_window_size

```

```python
print(longestMonotoneSublist(
    [1,3,
     1, 2, 3, 4, 8, 10, 14,
     3, 6, 8, 10, 15,
     1, 2, 3, 10, 16]))
print(longestMonotoneSublist([1, 2, 3, 4, 8, 10, 14]))

assert longestMonotoneSublist([1, 3, 2, 3, 4, 8, 10, 4, 8, 10]) == 5
assert longestMonotoneSublist(
    [1,3,
     1, 2, 3, 4, 8, 10, 14,
     3, 6, 8, 10, 15,
     1, 2, 3, 10, 16]) == 7
assert longestMonotoneSublist([1, 2, 3, 4, 8, 10, 14]) == 7
```

```python
def longestMonotoneSublist2(nums: [int]) -> int:

    # catches both [] and None
    if not nums:
        return 0

    if len(nums)==1:
        return 1

    left = 0
    right = 0
    max_window_size = 0

    while right <len(nums):
        # Cause I am a Java Guy - I am used to thinking
        # the for loop exist always gives the iterating
        # 'i' same as array.length() post-loop

        if (right>0 and nums[right] < nums[right-1]) or \
            right==len(nums)-1:
            # We are here to capture the baked 'window'
            # either because there was a dip, or
            # due to the right == last-index

            # Simply put, window-size depends on why we stopped
            # at 'right'!
            # If it was due to a dip, the window ended just
            # before the right-pointer!
            # Else, we are at the edge with right pointing to
            # the largest element of the current window -
            # the 'right' is part of the boundary!
            window_size = (right - left) if \
                nums[right] < nums[right-1] else \
                (right - left + 1)

            # Technically, this would be incorrect if the 'right' points
            # to the largest element of the current-window and had we
            # still not reached the edge of nums. But, that would never
            # happen. So we are good.
            left = right

            max_window_size = max(max_window_size, window_size)
        right += 1
    return max_window_size
```

```python
print(longestMonotoneSublist2(
    [1,3,
     1, 2, 3, 4, 8, 10, 14,
     3, 6, 8, 10, 15,
     1, 2, 3, 10, 16]))
print(longestMonotoneSublist2([1, 2, 3, 4, 8, 10, 14]))
print(longestMonotoneSublist2([1, 3,
                               2, 3, 4, 8, 10,
                               4, 8, 10]))
print(longestMonotoneSublist2([10,
                               1, 3, 4, 8,
                               4, 8, 10]))
print(longestMonotoneSublist2([10,
                               1, 3, 4, 8]))

print(longestMonotoneSublist2([10]))
print(longestMonotoneSublist2([10, 9, 8]))
print(longestMonotoneSublist2([10, 9, 8, 1, 3]))

# assert longestMonotoneSublist2([1, 3, 2, 3, 4, 8, 10, 4, 8, 10]) == 5
# assert longestMonotoneSublist2(
#     [1,3,
#      1, 2, 3, 4, 8, 10, 14,
#      3, 6, 8, 10, 15,
#      1, 2, 3, 10, 16]) == 7
# assert longestMonotoneSublist2([1, 2, 3, 4, 8, 10, 14]) == 7
```

```python
def test_sublist_logic(func):
    try:
        # 1. Boundary Cases: Empty and Single Elements
        assert func([]) == 0, f"Failed Empty List. Expected: 0, Got: {func([])}"
        assert func([7]) == 1, f"Failed Single Element. Expected: 1, Got: {func([7])}"

        # 2. The Trailing Sequence Bug (Tests if the final window is captured)
        assert func([10, 1, 2, 3, 4]) == 4, f"Failed Trailing Sequence. Expected: 4, Got: {func([10, 1, 2, 3, 4])}"

        # 3. The Decreasing Bug (Tests if the left pointer and max eval are correct)
        assert func([5, 4, 3, 2, 1]) == 1, f"Failed Strictly Decreasing. Expected: 1, Got: {func([5, 4, 3, 2, 1])}"

        # 4. Standard Mixed Arrays
        assert func([1, 2, 3, 4, 5]) == 5, f"Failed Strictly Increasing. Expected: 5, Got: {func([1, 2, 3, 4, 5])}"
        assert func([10, 1, 2, 3, 4, 2, 1]) == 4, f"Failed Middle Sequence. Expected: 4, Got: {func([10, 1, 2, 3, 4, 2, 1])}"

        # 5. Plateaus / Duplicates
        assert func([1, 2, 2, 3]) == 4, f"Failed Duplicates. Expected: 4, Got: {func([1, 2, 2, 3])}"

        print(f"Success! All test cases passed for {func.__name__}.")

    except AssertionError as e:
        print(e)

# --- How to run it ---
test_sublist_logic(longestMonotoneSublist2)
```
