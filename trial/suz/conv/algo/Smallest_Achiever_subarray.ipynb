```python

def smallest_achiever_subarray_incorrect(target:int, nums: [int]) -> int:
    '''

    :param nums:
    :param target:
    :return: -1 if the target could not be achieved
    '''
    if not nums:
        return -1 if target > 0 else 0

    if len(nums)==1:
        # INCORRECT LOGIC!
        return -1 if target < nums[0] else 1

    # for min, start with the max value
    min_window_size = -1
    left = 0
    current_window_sum = 0

    for right, val in enumerate(nums):
        current_window_sum += val

        # if the target > current win sum, continue
        # Else,

        ###! INCORRECT IMPLS:
        ####  (a) comparsion should be on current win sum, not val!
        ####  (b) 'left < right' skips the case where the single element
        #### achieves the target (i.e. left == right), and then fails to
        #### register to the min_window_size (because of the
        #### inequality check)
        while val >= target and left < right:
            # We are here,
            # means we found the state of 'target achieved'
            # at least once
            if(min_window_size == -1):
                min_window_size = 10000000

            current_window_size = right - left +1
            min_window_size = min(min_window_size, current_window_size)

            # Shrink the window until the target falls short again!
            current_window_sum -= nums[left]
            left += 1

    return min_window_size
```

```python

def smallest_achiever_subarray(target:int, nums: [int]) -> int:
    '''

    :param nums:
    :param target:
    :return: -1 if the target could not be achieved
    '''
    if not nums:
        return -1 if target > 0 else 0

    if len(nums)==1:
        # INCORRECT LOGIC!
        return -1 if target > nums[0] else 1

    # for min, start with the max value
    min_window_size = -1
    left = 0
    current_window_sum = 0

    for right, val in enumerate(nums):
        current_window_sum += val

        # if the target > current win sum, continue
        # Else,
        while target <= current_window_sum and left <= right:
            # We are here,
            # means we found the state of 'target achieved'
            # at least once
            if(min_window_size == -1):
                min_window_size = 10000000

            current_window_size = right - left +1
            min_window_size = min(min_window_size, current_window_size)

            # Shrink the window until the target falls short again!
            current_window_sum -= nums[left]
            left += 1

    return min_window_size
```
