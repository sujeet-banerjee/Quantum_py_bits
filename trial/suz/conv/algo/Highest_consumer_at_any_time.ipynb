#### Reporting Tool
Given:
  - List of persons by name (unique)
  - Corresp list of delta consumption (change in consumption event for the given person)

Find at any given time t, the *heighest* consumer so far. Consumption can be negative, which means he donated the liquor by the delta amount.

Example 1
---
- persons = ["101", "104", "101", "105"]
- consumption = [+4, +2, -4, +3]
- Expected: [4,4,2,3]

Example 2
---
- persons = ["150", "150", "120"]
- consumption = [+5, -5, 2]
- Expected: [5,0,2]


```python
persons = ["101", "104", "101", "105"]
consumption = [+4, +2, -4, +3]
#Expeccted = [4,4,2,3]
```

```python
import heapq

def report(persons:[str], consumption:[int]) -> [int]:
    aggregate = {}
    consumers_heap = []
    res = []
    for (person, delta) in zip(persons, consumption):
        # Make the consumption
        if not person in aggregate:
            aggregate[person] = 0
        aggregate[person] += delta

        # update the heap for sorting the aggregation
        heapq.heappush(consumers_heap, (-aggregate[person], person))

        #Reconcile - delete the stale entries
        # Peek at the top. If the top does not agree with the latest
        # consumption, dop it off the heap!
        while len(consumers_heap) > 0 and -consumers_heap[0][0] != aggregate[consumers_heap[0][1]]:
            # Keep popping
            heapq.heappop(consumers_heap)
        res.append(-consumers_heap[0][0])
    return res
```

```python
res = report(persons, consumption)
print(res)
```

```python
report(["150", "150", "120"], [+5, -5, +2])
```
