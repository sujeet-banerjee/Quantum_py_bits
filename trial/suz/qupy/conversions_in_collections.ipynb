## Dictinary Keys - immutability

```python
from torch.fx.immutable_collections import immutable_dict

#List to --> tuple, dict, set
# And back!

### List to various!
mlist = [1, 5, 6, 1, 2, 3]
print(mlist)

mTuple = tuple(mlist)
print(mTuple)

# DOES NOT WORK!
# mDict = dict(mlist)
# THIS DOES WORK!
mDict = dict(zip(mlist, mlist))
print(mDict)

mSet = set(mlist)
print(mSet)

# Back to list!
print("Set to List", list(mSet))
print("Tuple to List", list(mTuple))
print("Dict to List (restricted:Keys)", list(mDict.keys()))
print("Dict to List (restricted:Values)", list(mDict.values()))
print("Dict to List (Complete:Items)", list(mDict.items()))
```

```python
# Set to Frozenset
mFzSet = frozenset(mSet)
print(mFzSet)
print()
# Using Torch immutable collections
kwDict = dict(zip([1, 2, 3, 4, 5],
                  [(1, 'One'), (2, 'Two'), (3, 'Three'),
                   (4, 'Four'), (5, 'Five')]))
mFzDict = immutable_dict(kwDict)
print(mFzDict)
print("Retrieve: ", mFzDict[2])
print()

# Home grown -
# Note we convert the dict "items" to ordered key-value tuples
# and then make frozen set of the tuples, so the relative ordering
# of kw args does not matter!

mFzDict2 = frozenset(kwDict.items())
print(mFzDict2)
# Back to mutable dict
print(dict(mFzDict2))
print("Retrieve: ", dict(mFzDict2)[2])
```

```python
from collections import OrderedDict

#Dict

#Make a dict
my_dict = {"a": 1, "b": 2, "c": 3}
print(my_dict)
print(my_dict.keys())
print(my_dict.values())
print(my_dict.items())

new_dict = OrderedDict(my_dict.items())
new_dict.update({"t": 4})
new_dict.move_to_end('a', last=True)
print("Rearranged: ", new_dict)
print("Popped: ", new_dict.popitem())
print(new_dict)

'''
Output:
{'a': 1, 'b': 2, 'c': 3}
dict_keys(['a', 'b', 'c'])
dict_values([1, 2, 3])
dict_items([('a', 1), ('b', 2), ('c', 3)])
Rearranged:  OrderedDict({'b': 2, 'c': 3, 't': 4, 'a': 1})
Popped:  ('a', 1)
OrderedDict({'b': 2, 'c': 3, 't': 4})
'''
```
