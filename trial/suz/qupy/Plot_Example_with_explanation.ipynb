# 📈 Plotting Data with Matplotlib and NumPy
Welcome to this demonstration on plotting graphs in Python! In this notebook, we will use two of the most foundational libraries in data science: **NumPy** (for numerical data generation) and **Matplotlib** (for visualization).

### What does this code do?
This script simulates a common real-world scenario: capturing a continuous "true" signal that has been corrupted by random sensor noise. 

Here is a step-by-step breakdown of the operations performed in the code block below:

1. **Generating the X-Axis Data**:
   - We use `np.linspace(0, 2 * np.pi, 100)` to create an array of 100 evenly spaced numbers ranging from $0$ to $2\pi$ (approx. 6.28). This represents one full cycle of a wave.

2. **Creating the Clean Signal (The Ground Truth)**:
   - We pass our X-axis data into the sine function `np.sin(x)` to calculate the corresponding Y values. This generates a perfectly smooth mathematical sine wave.

3. **Simulating Real-World Noise**:
   - In reality, data is messy. We generate "noise" using `np.random.normal(0, 0.2, 100)`. This pulls 100 random numbers from a Gaussian (Normal) distribution with a mean (center) of `0` and a standard deviation (spread) of `0.2`.
   - We then physically add this noise array to our clean signal to create the corrupted `y_noisy` data.

4. **Visualization Engine**:
   - `plt.figure(...)`: Initializes the canvas size (10 inches wide by 5 inches tall).
   - `plt.plot(...)`: Draws the clean signal as a continuous blue line.
   - `plt.scatter(...)`: Draws the noisy data as individual, disconnected red dots on top of the blue line. The parameter `s=10` ensures these dots are relatively thin/small.
   - Finally, we render the grid, a legend to differentiate the data, and display the plot to the screen using `plt.show()`.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate 100 points between 0 and 2* pi
# np.linspace creates an evenly spaced sequence of numbers.
x = np.linspace(0, 2 * np.pi, 100)

# 2. Generate the 'Clean' Sine Wave
# Applies the mathematical sine function to every element in the 'x' array.
y_clean = np.sin(x)

# 3. Generate 'Noise' (Random values)
# Hint: Use np.random.normal(mean, standard_deviation, sample_size)
noise = np.random.normal(0, 0.2, 100)

# Corrupt the clean signal by adding the random noise to it.
y_noisy = y_clean + noise

# 4. Visualization
# Set the dimensions of the output graph (width=10, height=5)
plt.figure(figsize=(10, 5))

# Plot the continuous clean signal as a Line Graph
plt.plot(x, y_clean, label="Clean Signal", color="blue", linewidth=2)

# Overlay the corrupted signal as a Scatter Plot (individual dots)
# 's=10' controls the thickness of the scatter markers
plt.scatter(x, y_noisy,
            label="This will be shown as legend",
            color="red", s=10)

# Add chart metadata (title, legend, and background grid)
plt.title("Your Title: What is this graph about?")
plt.legend()
plt.grid(True)

# Render and display the plot
plt.show()
```
