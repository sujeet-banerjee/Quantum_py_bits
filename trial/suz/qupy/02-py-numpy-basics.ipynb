```python
# View the random Poisson Distribution!
import numpy as np
import matplotlib.pyplot as plt

val_pos = np.random.poisson(5, 1000)
plt.hist(val_pos, 80, facecolor='g', alpha=0.5)
plt.show()

```

```python
# View the random normal distribution!
val_pos = np.random.normal(80, 30, 1000)
plt.hist(val_pos, 50)
plt.show()
```
