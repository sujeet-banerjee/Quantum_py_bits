```python
# Dictionary

# str values
names = {
    'my' : 'Sujeet',
    'stdent': 'dsasdkjasdkj',
    'student' : 'Sumit',
    'father' : 'Anil',
    'uncle' : 'Sam',
    # duplicate values can be hazardous: 'my' is duplicate key
    # 'my' : 'Cat',
}

# float values
balances = {
    'sbi' : 3000000000.00,
    'icici': 19876543.00,
    'hdfc' : 20000000.00
}

# bool values
tasks = {
    'Get up from the bed': True,
    'brush my teeth': True,
    'had my lunch': False,
    'take classes': True,
    'buy grcy': False,
    'Create Assignments': True
}

# plain Set
ss = {'Sujeet', 'Sumit', 'Anil'}
```

```python
print(type(names))
print(type(balances))
print(type(tasks))
print(type(ss))
```

```python
print(names['student'])

# Incorrect index name
print(names['stdent'])

print(balances['sbi'])

print(type(balances['sbi']))
print(type(names['student']))
```

```python
# get all keys
print(tasks.keys())
print(balances.keys())
print(names.keys())

print(names['stdent'])

# no keys allowed for set, list, or tuple
# print(ss.keys())
```

```python
# get all values
print(tasks.values())
print(balances.values())
print(names.values())
```

```python
# Boolean 'in' operator to check if a given string
# is present as key or value.

if 'student' in names:
    print("'student' present in names as keys")

if 'student' in names.keys():
    print("'student' in names as keys")

# Incorrect key name 'stud'!
if 'Stud' in names.keys():
    print("'stud' in names as keys")
else:
    print("'stud' not in names as keys")

# Incorrect intent, as 'Sumit' is not a key, it's part of values!
if 'Sumit' in names:
    print("'Sumit' in names")
else:
    print("'Sumit' in not names as keys!")

if 'Sumit' in names.values():
    print("'Sumit' in names.values")
else:
    print("'Sumit' in not names.values")

```

```python
# Loop through the keys using for loop - example1
for the_key in names.keys():
    print("The value stored using the key: '" +
          the_key + "' is: "+ names[the_key])

# Loop through the keys - example2
print("\nI have accounts in these banks:")
for bnk_name_key in balances.keys():
    print("   -> "+ bnk_name_key)
```

```python
# Loop through the values using for loop - example1
balance = 0.0
for bnk_balance in balances.values():
    balance += bnk_balance
print("\nI have this total amount in various banks: " + str(balance))


# Loop through the values - example2
all_done = True
for task_done in tasks.values():
    all_done = all_done & task_done
print("\nHave I done all my taks? "+ str(all_done))
```

```python
# Loop through the keys and values - Example1
tasks_I_have_completed = ''
tasks_I_have_not_completed = ''
# The loop variable is a tuple: (task_name, task_status)
for (task_name, task_status) in tasks.items():
    if task_status:
        # Conditional: string-concat to 'completed'
        # only if the task_status == True
        tasks_I_have_completed += f",\n -> {task_name}"
    else:
        # else concat to 'not completed'
        tasks_I_have_not_completed += f",\n -> {task_name}"
print("\nTasks I have completed: "+ tasks_I_have_completed)
print("\nTasks I have NOT completed: "+ tasks_I_have_not_completed)
```
