```python
class Event:
    def __init__(self, date:str, type:str,
                 machine:str, user:str):
        self.date = date
        self.type = type
        self.machine = machine
        self.user = user
    def __str__(self):
        return f"{self.date}-{'L'+self.type[-1].\
            upper()}-{self.machine}-[{self.user}]"
```

```python
def print_events(events):
    print(" \n".join(map(str, events)))
    pass
```

```python
events = [
  Event('2020-01-21 12:45:46', 'login', 'myworkstation.local', 'jordan'),
  Event('2020-01-22 15:53:42', 'logout', 'webserver.local', 'jordan'),
  Event('2020-01-21 18:53:21', 'login', 'webserver.local', 'lane'),
  Event('2020-01-22 10:25:34', 'logout', 'myworkstation.local', 'jordan'),
  Event('2020-01-21 08:20:01', 'login', 'webserver.local', 'jordan'),
  Event('2020-01-23 11:24:35', 'login', 'mailserver.local', 'chris'),
]

print(events[0])
print_events(events)
```

```python
sorted_events = sorted(events, key=lambda e: e.date)
# print(events[1])
# print(events[2])
print_events(events)
print()
print_events(sorted_events)
```

```python
def create_machine_list(events:[Event]) -> dict:
    machines = {}
    sorted_events = sorted(events, key=lambda e: e.date)
    for event in sorted_events:
        if not event.machine in machines:
            machines[event.machine] = []
        if event.type == 'login':
            machines[event.machine].append(event.user)
        else:
            if event.user in machines[event.machine]:
                machines[event.machine].remove(event.user)

    return machines

machines = create_machine_list(events)
```

```python
    def generate_report(machines: dict):
    for machine, users in machines.items():
        if len(users) > 0:
            print(f"Machine: {machine} => users: {users}")

generate_report(machines)
```
