### SOlving System of LInear Equations
$$X_1 + X_2 − X_3 + 3X_4 = 4$$
$$3X_1 − X_2 + X_3 + 2X_4 = 2$$
$$−2X_1 + 0X_2 + X_3 + X_4 = 2$$
$$X_1 + 3X_2 − X_3 − X_4 = −4$$


```python
import numpy as np
```

Express the syst of equations as:
$$A.X = B$$

where A = [ \\
   1  1  -1 3 \\
   2  3  p \\
]

```python
# Define A matrix (coeff)
A = np.matrix([
  [1, 1, -1, 3 ],
    [3, -1, 1, 2 ],
    [-2, 0, 1, 1 ],
    [1, 3, -1, -1 ],
])
print(A)

# Define B matrix (value/outcome)
B = np.matrix([
  [4,2,2,-4 ]
]).reshape(4,1)
print(B)
```

```python
X = np.linalg.inv(A) @ B
print(X)
```

```python
E = B - (A@X)
print(E)
```

```python
E_sq = np.square(E)
print(E_sq)
```

```python
MSE = np.average(E_sq)
```

```python
RMSE = np.sqrt(MSE)
print("%f"%RMSE)
```
