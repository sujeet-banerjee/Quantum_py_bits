```python
import numpy as np
```

```python
# X = (A_inverse) @ B
'''
X = [
  x2,
  x3
]
'''

A = np.matrix([
    [2, 3],
    [3, 1]
])
print(A)

B = np.matrix([
    [8],
    [5]
])
print(B)
```

```python
# Inverse of A
A_inv = np.linalg.inv(A)
print(A_inv)
```

```python
# Validate correct invserse
print(A @ A_inv)
```

```python
# solve X
X = A_inv @ B
print(X)
```

```python
# Error calculation
B_predicted = A @ X
E = B - B_predicted


print(E)
```

```python
E_sq = np.square(E)
```

```python
MSE = np.average(E_sq)
```

```python
RMSE = np.sqrt(MSE)
print('{:f}'.format(MSE))
```
