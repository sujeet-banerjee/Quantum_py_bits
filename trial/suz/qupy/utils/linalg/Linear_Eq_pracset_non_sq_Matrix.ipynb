### SOlving System of LInear Equations
$$X_1 + X_2 + X_3 = 2$$
$$2X_1 − X_2 + X_3 = 5$$
$$−X_1 + 2X_2 + 2X_3 = 1$$
$$3X_1 − X_3 = 1$$
$$X_2 − 3X_3 = −7$$


```python
import numpy as np
```

Express the syst of equations as:
$$A.X = B$$
Where,
$ \mathbf{A} =
\begin{bmatrix}
      1 & 1 & 1 \\
      2 & -1 & 1 \\
      -1 & 2 & 2 \\
      3 & 0 & -1 \\
      0 & 1 & -3  \\
\end{bmatrix}$
, and
$\mathbf{B} =
\begin{bmatrix}
      2  \\
      5  \\
      1 \\
      1 \\
      -7  \\
\end{bmatrix}$


```python
# Define A matrix (coeff)
A = np.matrix([
  [1, 1,  1],
  [2, -1, 1],
  [-1, 2, 2],
  [3, 0, -1],
  [0, 1, -3],
]).reshape(5,3)
print(f"A={A}\n")

# Define B matrix (value/outcome)
B = np.matrix([
  [2,5,1,1,-7]
]).reshape(5,1)
print(f"B={B}\n")
```

### Treating non-Square matrix *A*
*Note:* We can't compute the inverse of a non-square matrix (A).
Thus, we multiply the A_transpose on both sides of:
$$A.X = B$$
to get:
$$A^T.A.X = A^T.B$$
or
$$A_{new}.X = B_{new}$$
where, $A_{new} = A^T.A$, and is a square matrix.

```python
# DOES NOT WORK
# X = np.linalg.inv(A) @ B
# print(X)
```

```python
A_new = A.transpose() @ A
print(A_new)

B_new = A.transpose() @ B
print(B_new)

X = np.linalg.inv(A_new) @ B_new
print(f"\nX={X}\n")
```

### The solution found is an approximation.
Recall that a matrix operation is also a projection.

The $A^T$ multiplied to $A$ has shrunken its dimentions from $5 \times 3$ to $3 \times 3$, and that is sort of a lossy compression (just like a shadow of a 3-D object is only a 2-D manifestation, which loses the 3D information).

Let's compute the MSE/RMSE by substituting the solution $X$ into the equations.

```python
E = B - (A@X)
print(f"{E}")
```

```python
E_sq = np.square(E)
print(E_sq)
```

```python
MSE = np.average(E_sq)
```

```python
RMSE = np.sqrt(MSE)
print("%f"%RMSE)
```

### Conclusion
Given the RMSE is almost 0.0, the solution obtained satisfies all the equations to the maximum possible extent.

However, this may not be true always (depending on the system of equations given in the question). For instance, the system below is not consistent (RMSE will be high). Find out as an exercise!
