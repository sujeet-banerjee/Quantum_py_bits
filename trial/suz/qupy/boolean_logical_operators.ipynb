```python
# The voter will be disqualified if
# he/she is below 18 years,
# OR,
# his/her name is not in the voter list
voter_list = ['Alex', 'Bob', 'Crysto', 'Don', 'Michael', 'Marry']


# Test if 'Marry' with age 16 will be not be allowed to vote;
(name, age) = ('Marry', 16)
if name not in voter_list or age < 18:
    print(name + " Cannot Vote.")  # Output: Marry Cannot Vote
else:
    print(name + " CAN vote!")

# Test if 'Phill' with age 23 will be not be allowed to vote;
(name, age) = ('Phill', 23)
if name not in voter_list or age < 18:
    print(name + " Cannot Vote.")  # Output: Phill Cannot Vote.
else:
    print(name + " CAN vote!")

# Test if 'Bob' with age 65 will be not be allowed to vote.
(name, age) = ('Bob', 65)
if name not in voter_list or age < 18:
    print(name + " Cannot Vote.")
else:
    print(name + " CAN vote!")    # Output: Bob CAN Vote!
```
