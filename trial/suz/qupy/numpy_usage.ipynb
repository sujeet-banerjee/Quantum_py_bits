```python
M1 = np.matrix(
    [
        [1, 2, 3],
        [4, 5, 6],
        [7, 8, 9]
    ]
)

print("M1 =\n" + str(M1))
print(f"Matrix order: {M1.shape}")

# Scalar multiplication
M2 = 5 * M1
print("M2 =\n" + str(M2))

# Matrix Addition
M3 = M1 + M2
print("M3 =\n" + str(M3))

# Matrix multiplication
M4 = M1 @ M2
print("M4 =\n" + str(M4))

# Dot product
print(np.dot([1, 2, 3] , [5, 20, 35]))
print(np.dot(M1[0,:], M2[:, 0]))

```

```python
# Example of a matrix with numbers from poisson distribution
M_Randm =np.random.randint(
    low=0,
    high=10, size=(3,3)
)
print(f"Random 3x3 matrix: \n{M_Randm}")
print("----------")

# Example of a matrix with numbers from poisson distribution
M_Poiss = np.random.poisson(lam=10, size=(3,3))
print(f"3x3 matrix of Poisson Distribution: \n{M_Poiss}")
```

```python
print(f"A 3x3 Poisson matrix: \n{M_Poiss}")
print()

# Row-wise or Column-wise operations using axis
max_indices_in_all_cols = np.argmax(M_Poiss, axis=0, keepdims=False)
print(f"Max indices in all columns: \n{max_indices_in_all_cols}")
print(M_Poiss.max(axis=0, keepdims=True))
print()

max_indices_in_all_rows = np.argmax(M_Poiss, axis=1, keepdims=False)
print(f"Max indices in all rows: \n{max_indices_in_all_rows}")
print(M_Poiss.max(axis=1, keepdims=True))

print()
# Row-wise average:
print(f"Row-wise Average: \n{M_Poiss.mean(axis=1, keepdims=True)}")


```

```python
# MAE and RMSE calculations using
## numpy (Method 1)
## Scikit-learn (Method 2)

import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Mock Data: Y (Actuals) and Y_hat (Predictions)
# Representing the housing price example (in $1000s)
Y = np.array([300, 450, 200, 600])
Y_hat = np.array([310, 430, 200, 620])

# ==========================================
# METHOD 1: Using Scikit-Learn (Recommended)
# ==========================================
mae_sklearn = mean_absolute_error(Y, Y_hat)

# Compute Mean Squared Error first, then take the square root.
# (Note: Scikit-learn 1.4+ also has a direct `root_mean_squared_error` function)
mse_sklearn = mean_squared_error(Y, Y_hat)
rmse_sklearn = np.sqrt(mse_sklearn)

print("--- Scikit-Learn Results ---")
print(f"MAE:  {mae_sklearn}")
print(f"RMSE: {rmse_sklearn:.2f}\n")


# ==========================================
# METHOD 2: Using Pure NumPy (Vectorized)
# ==========================================
# MAE: Mean of the absolute differences
mae_numpy = np.mean(np.abs(Y - Y_hat))

# RMSE: Square root of the mean of the squared differences
rmse_numpy = np.sqrt(np.mean((Y - Y_hat)**2))

print("--- NumPy Results ---")
print(f"MAE:  {mae_numpy}")
print(f"RMSE: {rmse_numpy:.2f}")
```
