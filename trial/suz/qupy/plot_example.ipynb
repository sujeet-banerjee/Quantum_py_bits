```python
import numpy as np
import matplotlib.pyplot as plt
# 1. Generate 100 points between 0 and 2* pi
x = np. linspace (0, 2 * np.pi , 100)
# 2. Generate the ’Clean ’ Sine Wave
y_clean = np.sin (x)
# 3. Generate ’Noise ’ ( Random values )
# Hint : Use np. random . normal (0, 0.2 , 100)
noise = np. random . normal (0, 0.2 , 100)
y_noisy = y_clean + noise
# 4. Visualization
plt. figure ( figsize =(10 , 5))
plt.plot (x, y_clean , label = "Clean Signal", color ="blue", linewidth =2)
plt. scatter (x, y_noisy , label ="dfdf", color ="red",
s =10)
plt. title ("sds")
plt. legend ()
plt. grid ( True )
plt. show ()
```
