#### MAX heap trick

```python
seq_2 = [1, -1, 2, 3]
#heapq.merge(heap_2, heap_1, key=lambda x: -x, reverse=True)
heap_2 = []
heap_reverse_2 = []
for elm in seq_2:
    # Min heap
    heapq.heappush(heap_2, elm)
    # Max heap
    heapq.heappush(heap_reverse_2, -elm)
print(heap_2)
print([elm for elm in map(lambda x: -x, heap_reverse_2)])
```

```python
import random
complex1: [()] = []
complex2: [()] = []
for elm in heap_2:
    heapq.heappush(complex1, (elm, random.randint(5, 53)))
    heapq.heappush(complex2, (random.randint(5, 53), elm))

'''
Expected output:

[(-1, 52), (1, 11), (2, 45), (3, 26)]
[(6, 2), (46, 3), (29, 1), (50, -1)]
'''

print(complex1)
print(complex2)


```

#### UNlike Java, there is no random 'delete'/'remove' supported in heapq
Therefore, to do the deletion optimally (in amortized linear time),
keep pushing, but deffer delete until pop. Use and external dict/map
to keep track of updated entries. While popping from the heap, ensure the top matches the value with the external dict!

#### Peek
Given this is sourced on a list, complex1[0] gives you a peek into the
heap!

```python
# Notice the := instead of =
# := is for create and assignment atomically
while( len(complex1) >0  and (item := heapq.heappop(complex1))):
    print(item)
```
