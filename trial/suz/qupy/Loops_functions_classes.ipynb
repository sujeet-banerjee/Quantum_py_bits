```python
# Methods

# Java
# String messsage(<params>) {
#     ....
#     return str;
# }
# Java
# sdsd() {
#
# }

# No input type needed (inferred at call-time)
# No output / return type need to be define (inferred)
def message(name):
    '''
    Just creates a salutation!
    :param name:
    :return: string with the salutation pre-pended!
    '''
    print("Received name: "+ str(name))
    return "Mr. " + str(name) + "!"

ret = message('Sujeet')
print(ret)



ret = message(True)
print(ret)
ret = message(1)
print(ret)

# ret = message2(True)
# print(ret)
```

```python
# Class definition
class Employee:
    # Java
    # private Str

    def __init__(self, input_name, loc='Delhi'):
        '''
        This is a contructor
        :param input_name:
        :param loc: Optional.
        '''

        # Public, can be corrupted
        self.name = input_name
        # Private, cannot be access
        self.__loc = loc
        print("Constructed employee: "+ str(self.name))

    def __str__(self):
        '''
        Similar to Java's toString()
        :return:
        '''
        return "Employee:"+self.name

    def get_name(self):
        return self.name

```

```python
# Instance
e1 = Employee("Sujeet", 'Bangalore')
print(e1.get_name())
print(e1.__loc)
print("-----------------")
e2 = Employee("Malancha")
e2.name = 'malancha'
print(e2.name)
e2.__loc = 'Bangalore'
print(e2.__loc)
```
