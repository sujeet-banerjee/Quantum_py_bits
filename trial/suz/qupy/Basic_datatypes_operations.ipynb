##### Why use float32?
A float64 takes up 8 bytes of memory, while a float32 takes up only 4 bytes. If you have an array of a billion numbers, switching to float32 instantly saves you 4 Gigabytes of RAM, at the cost of a few decimal places of extreme precision.

```python
import numpy as np

# Create a 32-bit float (Single precision)
var_32 = np.float32(3.14159)

# Create a 64-bit float (Double precision)
var_64 = np.float64(3.14159)

print(type(var_32))  # Output: <class 'numpy.float32'>
print(type(var_64))  # Output: <class 'numpy.float64'>
```

```python
#Tuples are immutable!
#They are ordered, often used to represent a "Relation"
# for example, a row in a database-Table.
t1 = (1, 2, 'a')
print(type(t1))
print(t1)

t2 = (1, 2, 'a')
print(type(t1))
print(t1)
print("checking t1 == t2: "+ str(t1 == t2))
print()

l1 = [2, 3, 1]
print(type(l1))
print(l1)

l2 = [1, 3, 2]
print(type(l2))
print(l2)
print("checking l1 == l2: "+ str(l1 == l2))
print()

# Order does not matter
s1 = {2, 3, 1}
print(type(s1))
print(s1)

s2 = {1, 2, 3}
print(type(s2))
print(s2)

print("checking s1 == s2: "+ str(s1 == s2))
```
