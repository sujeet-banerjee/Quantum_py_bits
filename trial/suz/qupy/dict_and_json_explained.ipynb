### 1. Defining Dictionaries and Sets
In this first block, we initialize several dictionaries. Notice how we can store different data types—strings, floats, and booleans—as values. 

**Key Concept:** A dictionary key must be unique and immutable. If you provide a duplicate key, the last one defined will overwrite the previous ones. We also define a **Set** (`ss`), which uses curly braces like a dictionary but contains only unique values without keys.

```python
# Dictionary

# str values
names = {
    'my' : 'Sujeet',
    'stdent': 'dsasdkjasdkj',
    'student' : 'Sumit',
    'father' : 'Anil',
    'uncle' : 'Sam',
}

# float values
balances = {
    'sbi' : 3000000000.00,
    'icici': 19876543.00,
    'hdfc' : 20000000.00
}

# bool values
tasks = {
    'Get up from the bed': True,
    'brush my teeth': True,
    'had my lunch': False,
    'take classes': True,
    'buy grcy': False,
    'Create Assignments': True
}

# plain Set
ss = {'Sujeet', 'Sumit', 'Anil'}
```

### 2. Verifying Data Types
It's important to confirm we are working with the correct objects. Even though dictionaries and sets both use curly braces `{}`, Python distinguishes them based on whether they contain key-value pairs (`dict`) or single items (`set`).

```python
print(type(names))
print(type(balances))
print(type(tasks))
print(type(ss))
```

### 3. Accessing Data via Keys
Unlike lists, where you use integer indices (0, 1, 2), dictionaries use their **keys** for access. This is much faster for large datasets because Python uses a hash table to find the value instantly without looping through the entire collection.

```python
print(names['student'])

# Incorrect index name
print(names['stdent'])

print(balances['sbi'])

print(type(balances['sbi']))
print(type(names['student']))
```

### 4. Exploring Dictionary Keys
The `.keys()` method returns a view object containing all the labels in the dictionary. This is useful when you need to check what categories of data are available or if you need to loop through the labels.

```python
# get all keys
print(tasks.keys())
print(balances.keys())
print(names.keys())

print(names['stdent'])
```

### 5. Retrieving Dictionary Values
Similarly, `.values()` gives us access to the data stored *under* those keys. This is particularly helpful when you want to perform calculations (like summing balances) without worrying about the labels themselves.

```python
# get all values
print(tasks.values())
print(balances.values())
print(names.values())
```

### 6. Membership Testing with 'in'
The `in` operator is a highly optimized way to check if a key exists before you try to access it, which helps prevent `KeyError` crashes. Note that by default, `if x in dictionary:` only checks the **keys**.

```python
# Boolean 'in' operator to check if a given string
# is present as key or value.

if 'student' in names:
    print("'student' present in names as keys")

if 'student' in names.keys():
    print("'student' in names as keys")

# Incorrect key name 'stud'!
if 'Stud' in names.keys():
    print("'stud' in names as keys")
else:
    print("'stud' not in names as keys")

# Incorrect intent, as 'Sumit' is not a key, it's part of values!
if 'Sumit' in names:
    print("'Sumit' in names")
else:
    print("'Sumit' in not names as keys!")

if 'Sumit' in names.values():
    print("'Sumit' in names.values")
else:
    print("'Sumit' in not names.values")

```

### 7. Iterating Over Keys
We can use a `for` loop to go through each key one by one. Once we have the key, we can use it to fetch and print the corresponding value from the dictionary.

```python
# Loop through the keys using for loop - example1
for the_key in names.keys():
    print("The value stored using the key: '" +
          the_key + "' is: "+ names[the_key])

# Loop through the keys - example2
print("\nI have accounts in these banks:")
for bnk_name_key in balances.keys():
    print("   -> "+ bnk_name_key)
```

### 8. Aggregation and Logic over Values
Dictionaries are perfect for aggregation. In these examples, we calculate a total bank balance and use boolean logic (`&`) to check if every task in our dictionary has been completed.

```python
# Loop through the values using for loop - example1
balance = 0.0
for bnk_balance in balances.values():
    balance += bnk_balance
print("\nI have this total amount in various banks: " + str(balance))


# Loop through the values - example2
all_done = True
for task_done in tasks.values():
    all_done = all_done & task_done
print("\nHave I done all my taks? "+ str(all_done))
```

### 9. Using .items() for Dual Iteration
The most efficient way to loop through a dictionary is using `.items()`. This returns both the key and the value simultaneously, allowing us to "unpack" them directly into variables within the loop header.

```python
# Loop through the keys and values - Example1
tasks_I_have_completed = ''
tasks_I_have_not_completed = ''
# The loop variable is a tuple: (task_name, task_status)
for (task_name, task_status) in tasks.items():
    if task_status:
        # Conditional: string-concat to 'completed'
        # only if the task_status == True
        tasks_I_have_completed += f",\n -> {task_name}"
    else:
        # else concat to 'not completed'
        tasks_I_have_not_completed += f",\n -> {task_name}"
print("\nTasks I have completed: "+ tasks_I_have_completed)
print("\nTasks I have NOT completed: "+ tasks_I_have_not_completed)
```

## 🌐 Processing JSON Data

JSON (JavaScript Object Notation) is the standard format for exchanging data on the web. In Python, JSON and Dictionaries are almost identical in structure, which makes it very easy to convert between them using the built-in `json` module.

### 10. Parsing JSON Strings
When you receive data from an API, it usually arrives as a **string**. We use `json.loads()` (load string) to convert that string into a Python dictionary so we can access it using keys.

```python
import json

# A JSON string representing user data
json_data = '{"id": 1, "name": "John Doe", "active": true, "roles": ["admin", "editor"]}'

# Convert JSON string to Python Dictionary
user_dict = json.loads(json_data)

print(f"Type: {type(user_dict)}")
print(f"User Name: {user_dict['name']}")
print(f"Primary Role: {user_dict['roles'][0]}")
```

### 11. Converting Dictionaries to JSON (Serialization)
If you want to send data to a web server or save it to a file, you must convert your dictionary back into a string. We use `json.dumps()` (dump string). 

**Pro Tip:** Use the `indent` parameter to make the resulting JSON string human-readable (pretty-printing).

```python
book_info = {
    "title": "Python 101",
    "author": "S. Banerjee",
    "available": True,
    "price": 29.99
}

# Convert Dictionary to JSON string with 4-space indentation
readable_json = json.dumps(book_info, indent=4)

print("Formatted JSON String:")
print(readable_json)
```
