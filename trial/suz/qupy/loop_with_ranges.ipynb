```python
# Repeated execution
print(1)
print(2)
print(3)
print(4)
print(5)
print(6)
print(7)
print(8)
print(9)
print(10)
```

```python
#Repeation (using 'for' loop)
for x in range(1, 11):
    print(x)

# All odd
for x in range(1, 11, 2):
    print(x)

# All Even
for x in range(2, 11, 2):
    print(x)

for x in range(11, 1, -1):
    print(x)
```

```python
#1+2+3+4+... + 21
total_till_now = 0

for x in range(1, 22):
    # Add each value (x), and store to the total ('=')
    total_till_now = total_till_now + x

print(total_till_now)
```
