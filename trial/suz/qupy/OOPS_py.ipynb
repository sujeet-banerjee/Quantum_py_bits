```python
import math


class Topic:
    '''
    For doc!!
    '''
    #Cons
    def __init__(self, name, complexity, hours=10):
        '''
        This must show up in the documentation??!!
        :param name:
        :param complexity:
        :param hours:
        '''

        self.name = name
        self.complexity = complexity
        self.hours = hours
    def __str__(self):
        return f"Topic: {self.name}[{self.complexity}]"

    # Notice: self and other!
    def __eq__(self, other):
        return other.name == self.name \
            and other.complexity == self.complexity
```

```python
quantum_phy_1 = Topic("Quantum Physics - I", 'E')
quantum_phy_1_back = Topic("Quantum Physics - I", 'E')

'''
Expected output:
----------------
Topic: Quantum Physics - I[E]
True
True
True
False
'''
print(quantum_phy_1)
print(quantum_phy_1 == quantum_phy_1)
print(quantum_phy_1 is quantum_phy_1)
print(quantum_phy_1 == quantum_phy_1_back)
print(quantum_phy_1 is quantum_phy_1_back)
```
