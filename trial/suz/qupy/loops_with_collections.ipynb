```python
students = [
    "James John Joules",
    "Sumit",
    "mary",
    "paul",
    "Aneesh",
    "sujeet",
    "Jaya"
]
```

```python
# what does the code do?
students[0]
```

```python
# what does the code do?
students[0][0]
```

```python
#print(type(students))
# Bad way
# print(students[0])
# print(students[1])
# print(students[2])

# correct way - use loop (All names)
for std in students:
    print(std[0])
```

```python
# print incorrect names (not starting with capital)
for std in students:
    if not std[0].isupper() :
        print(std)
```

```python
# print incorrect names (not starting with capital) - ANOTHER WAY!
for std in students:
    if std[0].islower() :
        print(std)
```

```python
# Fix the capitalization of names
# And address name with salutation, e.g. "Mr."
for std in students:
    print('Mr.'+ str.capitalize(std))

# WHAT IS THE PROBLEM WITH THE ADDRESSED NAMES?
```

```python
# To address gender-specific salutations, we need to have more data!
# We normally use tuples to associate related-details per entry - (name, gender, age, city, married?)
students = [
    ("James John Joules", 'M', 23, 'Toronto', True),
    ("Sumit", 'M', 31, 'Chennai', True),
    ("mary", 'F', 22, '', False),
    ("paul", 'M', 18, '', False),
    ("Aneesh", 'M', -100, '', False),
    ("sujeet", 'M', 47, 'Bangalore', True),
    ("Jaya", 'F', 31, 'Bangalore', True),
]
```

```python
# WHAT DO THESE CODE DO?
print(students[1][0][0])
print(students[1][0][1])
print(students[1][0][2])
print(students[1][0][3])
print(students[1][1])
print(students[1][2])
print(students[1][2] + 10)
print(students[1][3][0:4].upper())
print(students[1][3][0:4].lower())
print(not students[1][4])
```

```python
# Now printing with gender-specific salutations, names capitalized
for std in students:
    # condition within loop.
    if std[1] == 'M':
        print('Mr.'+ std[0].capitalize())
    else:
        print('Ms.'+ std[0].capitalize())
```

```python
# Similar code as above, but cleaner (use of more variables within the loop):
for std in students:
    gender = std[1]
    name = std[0].capitalize()
    # condition within loop.
    if gender == 'M':
        print('Mr.'+ name)
    else:
        print('Ms.'+ name)
```

```python
# WHAT DOES THIS CODE DO?
for std in students:
    gender = std[1]
    is_married = std[4]
    salutation = "Mr." if gender == 'M' else "Ms."
    salutation = "Mrs." if gender == 'F' and is_married else salutation
    name = salutation + ' '+ std[0].capitalize()
    message = "Welcome to my class."
    print(f"{name}! \n     {message} \n\n")
```
