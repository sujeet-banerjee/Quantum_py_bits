```python
from sqlalchemy.cyextension.resultproxy import operator
from tensorflow.compiler.tf2xla.python.xla import reduce

A = [1,2,3,4,5,6]
B = (1,2,3,4,5,6)
S = [ch for ch in "abcde"]
```

```python
print(A)
print(B)
print(S)

# For ZIP, the size is all that matters!
# You can zip tuples with list, etc
zipped_list = [(a,str(b),s) for a,b,s in zip(A,B, S)]
print(zipped_list)
string_zip = [sz for sz in zip(S,B)]
print(string_zip)
```

```python
def func(op:str, x:int, y:int) -> int:
    if op == '+':
        return x + y
    elif op == '*':
        return x * y
    elif op == '-':
        return x - y
    elif op == '%':
        return x % y
    else:
        return x / y

def func2(opcode) -> int:
    '''
    Takes the tuple (triple)
    and then disintgrates!
    :param opcode:
    :return:
    '''
    op, x, y = opcode
    if op == '+':
        return x + y
    elif op == '*':
        return x * y
    elif op == '-':
        return x - y
    elif op == '%':
        return x % y
    else:
        return x / y

# Zip
ops = [op for op in zip('+-*/%', A, B)]
print(ops)

# Unzip back into three lits
print([kw for kw in zip(*ops)])

print(*ops[0])

# DOes not work!
#reults = [res for res in map(func, ops)]
#print(reults)

# Works --> you need to disintegrate the tuple within
# the lambda of the function (func2)
reults = [res for res in map(func2, ops)]
print(reults)

# The same using lambda as a bridge to disintegrate
# the tuple, while using func again!
reults = [res for res in map(lambda opp : func(*opp) , ops)]
print(reults)
```

```python
# IMPORT!!!
from functools import reduce

# Reduce (adding example)
red = reduce(lambda x,y : x+y, A)
print(red)

# Reduce (add) without lambda
from operator import add, mul, floordiv, truediv, sub, pow
print(reduce(add, A))

# concat example
con = reduce(lambda red_val, elm: str(red_val)+str(elm), A)
print(con)

# conplex example
print(ops)
print(func2(ops[0]))
compl = reduce(lambda red_val, elm: red_val + func2(elm), ops, 0)
print(compl)
```
