### 1. Initializing the Data
We start by creating a simple list of strings called `students`. Notice that some names are capitalized correctly, while others are entirely lowercase.

```python
students = [
    "James John Joules",
    "Sumit",
    "mary",
    "paul",
    "Aneesh",
    "sujeet",
    "Jaya"
]
```

### 2. Accessing Elements
To access a specific item in the list, we use its index. Remember that Python lists are zero-indexed, so `0` represents the first item.

```python
# what does the code do?
students[0]
```

### 3. Chaining Indices
Because `students[0]` is a string, we can add a second index to access specific characters within that string. Here, we retrieve the first letter of the first name.

```python
# what does the code do?
students[0][0]
```

### 4. Basic Iteration
Instead of printing each item manually, we use a `for` loop to iterate through every student in the list and print the first character of their name.

```python
#print(type(students))
# Bad way
# print(students[0])
# print(students[1])
# print(students[2])

# correct way - use loop (All names)
for std in students:
    print(std[0])
```

### 5. Filtering with Conditionals
We can add an `if` statement inside our loop to find names that are not properly capitalized. Here, we use `not` combined with `.isupper()` to identify lowercase starts.

```python
# print incorrect names (not starting with capital)
for std in students:
    if not std[0].isupper() :
        print(std)
```

### 6. Alternative Filtering
This achieves the same goal as the previous block but uses the `.islower()` method for a more direct check.

```python
# print incorrect names (not starting with capital) - ANOTHER WAY!
for std in students:
    if std[0].islower() :
        print(std)
```

### 7. Transforming Data with Capitalize
Here, we attempt to fix the names and add a salutation. Take note of how `str.capitalize()` behaves on strings with multiple words like 'James John Joules'.

```python
# Fix the capitalization of names
# And address name with salutation, e.g. "Mr."
for std in students:
    print('Mr.'+ str.capitalize(std))

# WHAT IS THE PROBLEM WITH THE ADDRESSED NAMES?
```

### 8. Moving to Structured Data: Tuples
To handle gender-specific salutations and other details, we upgrade our list from simple strings to a list of **Tuples**. Each tuple represents a record containing multiple related attributes.

```python
# To address gender-specific salutations, we need to have more data!
# We normally use tuples to associate related-details per entry - (name, gender, age, city, married?)
students = [
    ("James John Joules", 'M', 23, 'Toronto', True),
    ("Sumit", 'M', 31, 'Chennai', True),
    ("mary", 'F', 22, '', False),
    ("paul", 'M', 18, '', False),
    ("Aneesh", 'M', -100, '', False),
    ("sujeet", 'M', 47, 'Bangalore', True),
    ("Jaya", 'F', 31, 'Bangalore', True),
]
```

### 9. Exploring Nested Indexing and Methods
We can perform complex operations by indexing into the list, then the tuple, and even into the strings within that tuple. This block demonstrates slicing, casing, and basic arithmetic.

```python
# WHAT DO THESE CODE DO?
print(students[1][0][0])
print(students[1][0][1])
print(students[1][0][2])
print(students[1][0][3])
print(students[1][1])
print(students[1][2])
print(students[1][2] + 10)
print(students[1][3][0:4].upper())
print(students[1][3][0:4].lower())
print(not students[1][4])
```

### 10. Logic Based on Tuple Data
Now we can use the data stored in the tuples (specifically index `1` for gender) to conditionally apply salutations like 'Mr.' or 'Ms.' during our loop.

```python
# Now printing with gender-specific salutations, names capitalized
for std in students:
    # condition within loop.
    if std[1] == 'M':
        print('Mr.'+ std[0].capitalize())
    else:
        print('Ms.'+ std[0].capitalize())
```

### 11. Refactoring for Readability
Code is often easier to read if we assign names to our indices. Here, we extract `gender` and `name` into variables before running the conditional logic.

```python
# Similar code as above, but cleaner (use of more variables within the loop):
for std in students:
    gender = std[1]
    name = std[0].capitalize()
    # condition within loop.
    if gender == 'M':
        print('Mr.'+ name)
    else:
        print('Ms.'+ name)
```

### 12. Final Application: Personalized Welcome Messages
In this final iteration, we use multiple tuple fields (gender and marital status) to create precise salutations and output a clean, multiline message using f-strings. Before executing, understand line-by-line, what the code does.

```python
# WHAT DOES THIS CODE DO?
for std in students:
    gender = std[1]
    is_married = std[4]
    salutation = "Mr." if gender == 'M' else "Ms."
    salutation = "Mrs." if gender == 'F' and is_married else salutation
    name = salutation + ' '+ std[0].capitalize()
    message = "Welcome to my class."
    print(f"{name}! \n     {message} \n\n")
```
