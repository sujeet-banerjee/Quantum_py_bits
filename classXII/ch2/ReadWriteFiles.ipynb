### Seek and Tell

```python
print("Learning to move the file object")
open("testfile.txt", "w+").close()
with open("testfile.txt", "r+") as f_handle:
    str = f_handle.read()
    print("Content: "+ str)
    print("Initially, the position of the file object is: ",
          f_handle.tell())
    f_handle.seek(0)
    print("Now the file object is at the beginning of the file:",
          f_handle.tell())
    f_handle.seek(10)
    print("We are moving to 10th byte position from the beginning of file")
    print("The position of the file object is at", f_handle.tell())
    str=f_handle.read()
    print(str)
```

### Pickle Module (Persistence)
* **dump** Dump serlized data to a file.
* **load** Load De-serlized data from a file, into the Python object.

```python
import pickle
listvalues=[
    (1,"Geetika",'F', 26, True),
    (1,"Sujeet",'M', 42, True),
    (1,"Alen",'M', 12, False),
]
fileobject=open("mybinary.dat", "wb+")
pickle.dump(listvalues, fileobject)
fileobject.close()
```

```python
import pickle
print("The data that were stored in file are: ")
fileobject=open("mybinary.dat","rb")
objectvar=pickle.load(fileobject)
fileobject.close()

### Print the inflated contents and types
print(objectvar)
print(type(objectvar))
print(type(objectvar[0]))
```
