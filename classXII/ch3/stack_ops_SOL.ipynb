# Stack and Operations

````python
st = list()

# Push adds the element at the end.
st.append('x')
st.append('y')
st.append('z')
print("Stack: "+ str(st))

# Size of the stack
print("Stack-size: "+ str(len(st)))

# Top element / Peek
print("Stack-top: "+ str(st[-1]))
````

````python
# Pop until stack is empty
while len(st) != 0:
    elm = st.pop()
    print(f"Popped element: {elm}")
````

## Uses of Stack
* **Parenthesis Matching** - Say, ```'(()(()))'``` is correct while ```'(()()))('``` is incorrect, though both expressions have equal number of opening and closing parenthesis.
* **Pattern-Matching** - Find all code-words that have all '**a**'s before '**c**' such that the count of '**a**' is same as that of '**c**'. Example, ```'bestanatoacicc'``` is a valid string as the three **a**s occur before the three **b**s.

````python
# Parenthesis Matching example

## Test data
test_string_correct = '(3+(2-3)*(3*(5-4))-2)'
test_string_incorrect = '(3+(2-3)*(5-4))-2)'

def validate_paren_match(inp:str) -> bool:
    if len(inp) == 0:
        return True

    ## empty list as stack
    paren_stack = list()

    for ch in inp:
        if ch == '(':
            # PUSH
            paren_stack.append(ch)
        elif ch == ')':
            # Pop and match
            if len(paren_stack) == 0:
                # MISMATCH
                return False
            else:
                closing = paren_stack.pop()
        else:
            # Ignore if the character is not a paranth.
            pass

    # If all parenth. matched, the stack should be empty!!
    return len(paren_stack) == 0
````

### Mathematical Notations in Computer Science

#### 1. Infix Notation
**Definition:** The standard mathematical and logical notation where the operator is placed *in between* the operands it acts upon.

**Example:** `3 + 4` (The `+` is the operator, between operands `3` and `4`)

##### Key Characteristics:
* **Human Readability:** This is the universal language of mathematics taught to humans. It mirrors natural language (e.g., "three plus four") and is highly intuitive.
* **Programming Standard:** Most modern programming languages (Python, Java, C++) use infix notation for mathematical logic (e.g., `result = (a * b) + c`).
* **Binary Context:** It acts as a clear visual pivot point separating the left-hand side from the right-hand side in binary operations (AND, OR, +, -, *, /).

##### The "Flaw" (Machine Ambiguity):
Infix notation is inherently ambiguous to computers without additional rules.
For the expression `2 + 3 * 4`, a computer must be programmed with:
1. **Order of Operations (PEMDAS/BODMAS):** To know that multiplication happens before addition.
2. **Parentheses:** To override standard precedence (e.g., `(2 + 3) * 4`).
Because of this overhead, computers typically translate human-written infix notation into abstract syntax trees and postfix/prefix notation before evaluating it.

---

#### 2. Prefix Notation (Polish Notation)
**Definition:** A notation invented by Jan Łukasiewicz where the **operator comes *before* its operands**.

**Example:** Instead of `3 + 4`, you write `+ 3 4`.

##### Key Characteristics:
* **Zero Ambiguity:** Prefix notation entirely eliminates the need for order of operations rules and parentheses. Every expression has exactly one valid way to be read.
    * *Infix:* `2 + (3 * 4)`  ➔ *Prefix:* `+ 2 * 3 4`
    * *Infix:* `(2 + 3) * 4`  ➔ *Prefix:* `* + 2 3 4`
* **Computational Efficiency:** Computers can evaluate prefix expressions in a single, fast pass (often reading right-to-left) using a simple Stack data structure, saving memory and processing time.

##### Primary Use Cases:
* **Compilers & ASTs:** When a compiler reads code, it translates infix notation into an Abstract Syntax Tree (AST). Prefix notation is a direct, flattened textual representation of this tree.
* **Functional Programming:** Languages like LISP, Scheme, and Clojure rely heavily on prefix notation. In LISP, everything is a function call, making the syntax highly uniform: `(+ 1 2 3)` or `(function arg1 arg2)`.

## Convert to postfix - Part A
#### No Paranthesis

````python
def isOp(ch):
    return ch == '+' or ch == '-' or ch == '*' or ch == '/'

isOp('+')
isOp('x')
````

````python
def precedes(first, second) -> bool:
    '''
    CHecks if the ``first`` operator has higher precedence than
    the ``second`` one.
    :param first:
    :param second:
    :return: True if first has heigher precd. than the second. False otherwise.
    '''
    if first == '/':
        return True
    elif first == '*':
        return second == '+' or second == '-'
    else:
        return False
````

````python
# Read Input from the user
inp = input("Input an expression")
print(f"INput: {inp}")
st.clear()
output = []
for ch in inp:
    if isOp(ch):
        if len(st) !=0 and precedes(st[-1], ch):
            output.append(st.pop())
            st.append(ch)
        else:
            st.append(ch)
            print("PUSH: %s" % st)
    else:
        output.append(ch)
print(st)
st.reverse()
output.extend(st)
print(f"{output}")
````

**Note:** The bottom of the stack has the least precedence operator. The precedence drops as you go down the stack from the top.

### Introduce parenthesis in the INfix expression

* Treat the expression between '(' and ')' independently.

````python
def infix_to_postfix(inp:str)->[]:
    print(f"INput: {inp}")
    st = list()
    output = []
    for index, ch in enumerate(inp):
        if isOp(ch):
            if len(st) !=0 and precedes(st[-1], ch):
                output.append(st.pop())
                st.append(ch)
            else:
                st.append(ch)
                print("PUSH: %s" % st)
        elif ch == '(':
            print("Pushed '(' to the stack")
            st.append(ch)
        elif ch == ')':
            # Pop everything until the last '(' from the stack
            top = st.pop()
            while top != '(':
                output.append(top)
                top = st.pop()
        else:
            output.append(ch)
    print(st)
    st.reverse()
    output.extend(st)
    print(f"{str(output)}")
    return output
````

### Evaluating Postfix expression

````python
def apply(op, x, y):
    x = int(x)
    y = int(y)
    if op == '*':
        return x * y
    elif op == '+':
        return x + y
    elif op == '-':
        return x - y
    elif op == '/':
        return x / y
    else:
        raise Exception("Unknown operator: "+ str(op))

def eval_postfix(inp:[]):
    st = []
    for ch in inp:
        if isOp(ch):
            if len(st) == 0:
                raise Exception("Invalid State: No operads left")
            else:
                y = st.pop()
                x = st.pop()
                val = apply(ch, x, y)
                st.append(val)
        else:
            st.append(ch)
    if len(st) != 1:
        raise Exception("Invalid State: Invalid postfix string")
    return st[-1]
````

````python
res = eval_postfix(postfix)
print(f"Value of '{postfix}' = {res} ")
````
